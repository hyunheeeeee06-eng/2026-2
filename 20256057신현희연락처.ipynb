```python

contacts = []

# 연락처 추가
def add_contact():
    name = input("이름: ")
    phone = input("전화번호: ")
    address = input("주소: ")
    birthday = input("생년월일: ")
    memo = input("기타 메모: ")

    contact = {
        "이름": name,
        "전화번호": phone,
        "주소": address,
        "생년월일": birthday,
        "메모": memo
    }

    contacts.append(contact)
    print("연락처를 추가했습니다.")

# 연락처 삭제
def delete_contact():
    name = input("삭제할 이름: ")

    for contact in contacts:
        if contact["이름"] == name:
            contacts.remove(contact)
            print("연락처를 삭제했습니다.")
            return

    print("연락처를 찾을 수 없습니다.")

# 연락처 검색
def search_contact():
    name = input("검색할 이름: ")

    for contact in contacts:
        if contact["이름"] == name:
            print(contact)
            return

    print("연락처를 찾을 수 없습니다.")

# 연락처 수정
def update_contact():
    name = input("수정할 이름: ")

    for contact in contacts:
        if contact["이름"] == name:
            contact["전화번호"] = input("새 전화번호: ")
            contact["주소"] = input("새 주소: ")
            contact["생년월일"] = input("새 생년월일: ")
            contact["메모"] = input("새 메모: ")

            print("연락처를 수정했습니다.")
            return

    print("연락처를 찾을 수 없습니다.")

# 연락처 전체 출력
def print_contacts():
    if len(contacts) == 0:
        print("저장된 연락처가 없습니다.")
    else:
        for contact in contacts:
            print(contact)

# 메뉴
def menu():
    while True:
        print("\n========== menu ==========")
        print("1. 연락처 추가")
        print("2. 연락처 삭제")
        print("3. 연락처 검색")
        print("4. 연락처 수정")
        print("5. 연락처 전체 출력")
        print("6. 종료")
        print("==========================")

        choice = input("번호를 선택하시오: ")

        if choice == "1":
            add_contact()
        elif choice == "2":
            delete_contact()
        elif choice == "3":
            search_contact()
        elif choice == "4":
            update_contact()
        elif choice == "5":
            print_contacts()
        elif choice == "6":
            print("프로그램을 종료합니다.")
            break
        else:
            print("잘못된 번호입니다.")

menu()
```
